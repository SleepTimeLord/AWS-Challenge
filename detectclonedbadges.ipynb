```python
# old system
# gets the difference between the first and second unique loc
def diff_between_unique_place(group, place):
    #saves the first min and min location
    firstMin = group["timestamp"].min()
    firstLoc = group.loc[group["timestamp"].idxmin(), place]

    # filters out same location 
    filteredGroup = group[group[place] != firstLoc] 

    # finds second min
    secondMin = filteredGroup["timestamp"].min()

    return secondMin - firstMin
```

```python
# place has to be building_id, room_id, or location_id
def smallest_diff_between_unique_place(group, place):

    diffBetweenUniquePlace = []

    sortedValues = group.sort_values(by="timestamp").reset_index(drop=True)

    for i in range(len(sortedValues) - 1):
        # if location is the same go next because its not unique location
        if sortedValues.loc[i, place] == sortedValues.loc[i+1, place]:
            continue

        # get the difference between loc
        diff = sortedValues.loc[i+1, "timestamp"] - sortedValues.loc[i, "timestamp"]

        diffBetweenUniquePlace.append(diff)

    if not diffBetweenUniquePlace:
        return "NaT"
    else:
        return min(diffBetweenUniquePlace)
```

```python
# groups by the user so its all the unique users and it applys the difference_between_unique_loc function
locDiff = df.groupby("user_id").apply(lambda group: smallest_diff_between_unique_place(group, "location_id"), include_groups=False)

locFilterNaT = locDiff[locDiff != "NaT"]
# get users if travel times between 2 location is less than 4 hours 
filterLoc = locFilterNaT[locFilterNaT < pd.Timedelta("4h")]

clonedUserBadges= filterLoc.reset_index()["user_id"]

clonedUserBadges
```

```python
# groups the users and does the same thing on the top
roomDiff = df.groupby("user_id").apply(lambda group: smallest_diff_between_unique_place(group, "room_id"), include_groups=False)

roomFilterNaT = roomDiff[roomDiff != "NaT"]
# filter the buildingDiff series its duplicate time
clonedUserBadges = roomFilterNaT[(roomFilterNaT < pd.Timedelta("100ms"))]
#clonedUserBadges
clonedUserDF = clonedUserBadges.to_frame().reset_index()
clonedUserDF

# get succes == False
filterByFailedEntry = df[df["success"] == False]
filterByFailedEntry

# get only unique values
curiousUsers = filterByFailedEntry["user_id"].unique()

print(f"Amount of users that have failed to access rooms that they are not authorized to access {len(curiousUsers)}")
clonedUserDF[clonedUserDF["user_id"].isin(curiousUsers)]
```
